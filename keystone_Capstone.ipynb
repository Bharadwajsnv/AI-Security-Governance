```python
print("=" * 70)
print("SCENARIO 8: ENCRYPT SYNTHETIC CUSTOMER PII")
print("=" * 70)
print("Purpose: Verify that sensitive data is not stored")
print("in readable plaintext form.")
print()

sensitive_customer = SensitiveCustomerData(
    application_id="KS-001",
    national_insurance_number="QQ123456C",
    bank_account_number="12345678",
    sort_code="11-22-33",
    credit_score=780
)

encryption_service = EncryptionService()

encrypted_customer_record = (
    encryption_service.encrypt_customer_data(
        sensitive_customer
    )
)

print("APPLICATION ID:")
print(encrypted_customer_record["application_id"])

print()
print("ENCRYPTED NATIONAL INSURANCE NUMBER:")
print(encrypted_customer_record["national_insurance_number"])

print()
print("ENCRYPTED BANK ACCOUNT NUMBER:")
print(encrypted_customer_record["bank_account_number"])

print()
print("ENCRYPTED SORT CODE:")
print(encrypted_customer_record["sort_code"])

print()
print("FINAL STATUS: SENSITIVE DATA ENCRYPTED")
```

```python
print("=" * 70)
print("SCENARIO 9: AUTHORISED UNDERWRITER PII ACCESS")
print("=" * 70)
print("Purpose: Verify that an authorised role can access")
print("protected information for legitimate underwriting.")
print()

secure_data_service = SecureDataAccessService(
    encryption_service
)

authorised_access_result = (
    secure_data_service.request_protected_data(
        encrypted_record=encrypted_customer_record,
        requesting_role=ROLE_UNDERWRITER
    )
)

print("REQUESTING ROLE:", authorised_access_result.role)
print("ACCESS STATUS:", authorised_access_result.status)
print("REASON:", authorised_access_result.reason)

if authorised_access_result.data:
    print()
    print("DECRYPTED DATA AVAILABLE TO AUTHORISED ROLE")
    print("-" * 70)
    print(
        "Application ID:",
        authorised_access_result.data["application_id"]
    )
    print(
        "National Insurance Number:",
        authorised_access_result.data[
            "national_insurance_number"
        ]
    )
    print(
        "Bank Account Number:",
        authorised_access_result.data[
            "bank_account_number"
        ]
    )
    print(
        "Sort Code:",
        authorised_access_result.data["sort_code"]
    )
    print(
        "Credit Score:",
        authorised_access_result.data["credit_score"]
    )
```

```python
print("=" * 70)
print("SCENARIO 10: UNAUTHORISED PII ACCESS ATTEMPT")
print("=" * 70)
print("Purpose: Verify that a role without permission cannot")
print("access protected customer information.")
print()

unauthorised_access_result = (
    secure_data_service.request_protected_data(
        encrypted_record=encrypted_customer_record,
        requesting_role=ROLE_CUSTOMER_SERVICE
    )
)

print("REQUESTING ROLE:", unauthorised_access_result.role)
print("ACCESS STATUS:", unauthorised_access_result.status)
print("REASON:", unauthorised_access_result.reason)
print("PROTECTED DATA RETURNED:", unauthorised_access_result.data)

if unauthorised_access_result.data is None:
    print()
    print("FINAL STATUS: ACCESS DENIED - NO PII DISCLOSED")
```

```python
print("=" * 70)
print("SCENARIO 11: PII-REDACTED AUDIT LOGGING")
print("=" * 70)
print("Purpose: Verify that sensitive identifiers are removed")
print("before application activity is written to the audit log.")
print()

audit_log = TamperEvidentAuditLog()

sensitive_log_message = (
    "Application received with NI number QQ123456C, "
    "bank account 12345678 and sort code 11-22-33."
)

audit_event_1 = audit_log.add_event(
    event_type="APPLICATION_RECEIVED",
    application_id="KS-001",
    message=sensitive_log_message,
    actor_role="KEYSTONE_AGENT"
)

print("ORIGINAL MESSAGE:")
print(sensitive_log_message)

print()
print("STORED AUDIT MESSAGE:")
print(audit_event_1["message"])

print()
print("AUDIT EVENT ID:", audit_event_1["event_id"])
print("AUDIT EVENT HASH:", audit_event_1["event_hash"])
```

```python
audit_log.add_event(
    event_type="M3_SECURITY_CHECK",
    application_id="KS-001",
    message="Prompt-injection check completed. No attack detected.",
    actor_role="INJECTION_GUARD"
)

audit_log.add_event(
    event_type="M4_POLICY_DECISION",
    application_id="KS-001",
    message="Application evaluated within the £500,000 policy boundary.",
    actor_role="ACTION_POLICY_ENGINE"
)

audit_log.add_event(
    event_type="DIP_DECISION",
    application_id="KS-001",
    message="Automatic DIP processing permitted by policy.",
    actor_role="KEYSTONE_AGENT"
)

print()
print("TOTAL AUDIT EVENTS:", len(audit_log.events))
```

```python
print("=" * 70)
print("SCENARIO 12A: VERIFY UNMODIFIED AUDIT CHAIN")
print("=" * 70)

integrity_result = audit_log.verify_integrity()

print("AUDIT CHAIN VALID:", integrity_result["valid"])
print("RESULT:", integrity_result["reason"])

if integrity_result["valid"]:
    print("FINAL STATUS: AUDIT INTEGRITY VERIFIED")
```

```python
print("=" * 70)
print("SCENARIO 12B: AUDIT LOG TAMPERING ATTACK")
print("=" * 70)
print("Purpose: Deliberately modify a historical event and")
print("verify that the integrity check detects the change.")
print()

original_message = audit_log.events[1]["message"]

audit_log.events[1]["message"] = (
    "Security check was bypassed and no alert was generated."
)

tampered_integrity_result = audit_log.verify_integrity()

print("MODIFIED EVENT:", audit_log.events[1]["event_id"])
print("ORIGINAL MESSAGE:", original_message)
print("TAMPERED MESSAGE:", audit_log.events[1]["message"])

print()
print("AUDIT CHAIN VALID:", tampered_integrity_result["valid"])
print("FAILED EVENT:", tampered_integrity_result["event_id"])
print("RESULT:", tampered_integrity_result["reason"])

if not tampered_integrity_result["valid"]:
    print()
    print("FINAL STATUS: AUDIT TAMPERING DETECTED")
```

```python
audit_log = TamperEvidentAuditLog()

audit_log.add_event(
    event_type="APPLICATION_RECEIVED",
    application_id="KS-001",
    message="Synthetic mortgage application received.",
    actor_role="KEYSTONE_AGENT"
)

audit_log.add_event(
    event_type="M3_SECURITY_CHECK",
    application_id="KS-001",
    message="Prompt-injection check completed.",
    actor_role="INJECTION_GUARD"
)

audit_log.add_event(
    event_type="M4_POLICY_DECISION",
    application_id="KS-001",
    message="Action-authority policy evaluated.",
    actor_role="ACTION_POLICY_ENGINE"
)

print("Clean audit chain recreated.")
print(audit_log.verify_integrity())
```
